## 5. Open an IFC file

Place an IFC file in the notebook's working folder (or provide the correct relative or absolute path). 

- Replace `example.ifc` with your actual filename.

**Something that can be tricky:** A ZIP download from GitHub must be extracted before you can use the files inside it. The notebook's working directory may also differ from the folder containing the `.ipynb` file.

```python
ifc_path = Path("example.ifc")  # Replace with your IFC filename (if the file is located in the folder) or with full path
model = None

if not ifc_path.is_file():
    print(f"IFC file not found: {ifc_path.resolve()}")
    print("Check the filename, folder, or path and try again.")
else:
    try:
        model = ifcopenshell.open(str(ifc_path))
        print(f"Loaded IFC model: {ifc_path.name}")
    except Exception as error:
        print(f"Could not open the IFC file: {error}")
```

## 6. Find building elements

IFC classes describe different kinds of building elements and their placement of properties. 

In this example we count walls, windows and doors.

```python
if model is not None:
    for ifc_class in ["IfcWall", "IfcWindow", "IfcDoor"]:
        elements = model.by_type(ifc_class)
        print(f"{ifc_class}: {len(elements)}")
else:
    print("Load an IFC file in step 3 first.")
```

**Try it yourself:** Change `IfcDoor` to `IfcSlab` and run the cell again. The numbers will depend on your IFC model.

## 7. Read information from one element

This example shows the name, IFC class and GlobalId of the first wall found. The exact values depend on the model you choose.

```python
if model is not None:
    walls = model.by_type("IfcWall")
    if walls:
        wall = walls[0]
        print("Name:", wall.Name or "Not specified")
        print("IFC class:", wall.is_a())
        print("GlobalId:", wall.GlobalId)
    else:
        print("No IfcWall elements found in this model.")
else:
    print("Load an IFC file in step 3 first.")
```

**Try it yourself:** Replace `IfcWall` with `IfcWindow` in the cell above.
